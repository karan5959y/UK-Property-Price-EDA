```python
Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower:", lower)
print("Upper:", upper)
```

```python
outliers = df[df["Price"] > upper]

print("Number of outliers:", len(outliers))
print("Percentage:", len(outliers) / len(df) * 100)
```

```python
city_property = pd.crosstab(df["Town_City"], df["Property_Type"])

city_property["Top_Property_Type"] = city_property.idxmax(axis=1)
city_property["Top_Transactions"] = city_property[["D", "F", "O", "S", "T"]].max(axis=1)

city_property.sort_values("Top_Transactions", ascending=False).head(10)
```

```python
city_newbuild = pd.crosstab(
    df["Town_City"],
    df["New_Build"],
    normalize="index"
) * 100

city_newbuild["New_Build_%"] = city_newbuild["Y"]

city_newbuild.sort_values("New_Build_%", ascending=False).head(10)
```

```python
city_newbuild = pd.crosstab(
    df["Town_City"],
    df["New_Build"],
    normalize="index"
) * 100

city_count = df["Town_City"].value_counts()

city_newbuild["New_Build_%"] = city_newbuild["Y"]

result = city_newbuild[city_count >= 500] \
    .sort_values("New_Build_%", ascending=False) \
    .head(10)

result
```
